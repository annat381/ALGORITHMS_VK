```python
#1
def reverse(arr, left, right):
    while left < right:
        arr[left], arr[right] = arr[right], arr[left]
        left += 1
        right -= 1

def rotate(arr, k):
    n = len(arr)
    if n == 0:
        return arr
    k %= n                  # если k > n
    reverse(arr, 0, n - 1)
    reverse(arr, 0, k - 1)
    reverse(arr, k, n - 1)
    return arr

arr=[1, 2, 3, 4, 5, 6, 7]
k=3 
rotate(arr,k)
print(arr)
```

```python
#2
def merge_sorted_arrays(arr1, arr2):
    merged = []
    i = j = 0
    while i < len(arr1) and j < len(arr2):
        if arr1[i] < arr2[j]:
            merged.append(arr1[i]); i += 1
        else:
            merged.append(arr2[j]); j += 1
    # один из массивов закончился, остаток дописываем
    merged.extend(arr1[i:])
    merged.extend(arr2[j:])
    return merged

```

```python
#3без аллокаций 
def merge(arr1, arr2):
    p1 = len(arr1) - len(arr2) - 1
    p2 = len(arr2) - 1
    p3 = len(arr1) - 1

    while p2 >= 0:
        if p1 >= 0 and arr1[p1] > arr2[p2]:
            arr1[p3] = arr1[p1]
            p1 -= 1
        else:
            arr1[p3] = arr2[p2]
            p2 -= 1
        p3 -= 1
    return arr1


```

Условие цикла while p2 >= 0. Когда arr2 закончился, остаток arr1 уже стоит на своих местах, ничего копировать не нужно.
Проверка p1 >= 0 защищает от выхода за границу, когда элементы arr1 кончились раньше.
Запись никогда не затирает непрочитанные данные: p3 >= p1 всегда.

```python
#4 0 and 1 
def sort_binary_array(arr):
    left, right = 0, len(arr) - 1
    while left < right:
        if arr[left] == 0:
            left += 1
        elif arr[right] == 1:
            right -= 1
        else:  # arr[left] == 1 и arr[right] == 0
            arr[left], arr[right] = arr[right], arr[left]
            left += 1
            right -= 1
    return arr

arr=[0,1,1,0,1,0,1,0]
sort_binary_array(arr)
print(arr)


```

```python
#5  0 1 and 2 flag 
def sort_colors(nums):
    low, mid, high = 0, 0, len(nums) - 1
    while mid <= high:
        if nums[mid] == 0:
            nums[low], nums[mid] = nums[mid], nums[low]
            low += 1
            mid += 1
        elif nums[mid] == 1:
            mid += 1
        else:
            nums[mid], nums[high] = nums[high], nums[mid]
            high -= 1
    return nums


```

Самое важное: почему при 0 двигаем mid, а при 2 нет.

При 0: на позиции low может стоять только 1 (или сам mid), потому что зона [low, mid) состоит из единиц. После обмена на mid оказывается уже проверенная единица, можно идти дальше.
При 2: с high приходит непроверенный элемент, он может быть 0, 1 или 2. Его надо проверить, поэтому mid остаётся на месте.

```python
#6 четные вперед 


def even_first(arr):
    even_index = 0
    for i in range(len(arr)):
        if arr[i] % 2 == 0:
            arr[i], arr[even_index] = arr[even_index], arr[i]
            even_index += 1
    return arr
arr=[7,3,2,4,1,11,8,9]
even_first(arr)
print(arr)

```

При  evens + odds порядок внутри каждой группы не нарушается. Ключевой принцип стабильности: мы никогда не меняем местами два элемента одной группы (оба чётных или оба нечётных) — мы лишь разделяем их по группам, сохраняя исходную последовательность внутри каждой.

```python
#7 0 in the end 

def move_zeroes(arr):
    write = 0
    for i in range(len(arr)):
        if arr[i] != 0:
            arr[i], arr[write] = arr[write], arr[i]
            write += 1
    return arr
arr=[0,0,1,0,3,12]
move_zeroes(arr)
print(arr)
```
