```python
def binary_search_sqrt(target: int) -> int:
    if target < 0:
        raise ValueError("<0")
    if target == 0 or target == 1:
        return target
    l, r = 0, target
    ans = 0
    while l <= r:
        mid = (l + r) // 2
        if mid * mid == target:
            return mid
        elif mid * mid < target:
            ans = mid
            l = mid + 1
        else:
            r = mid - 1
    return ans

print("binary_search_sqrt(15) =", binary_search_sqrt(15))


def copy_time(n: int, x: int, y: int) -> int:
    if n == 0:
        return 0
    if n == 1:
        return min(x, y)
    l, r = 0, min(x, y) * n
    ans = r
    while l <= r:
        mid = (l + r) // 2
        if (mid // x) + (mid // y) >= n:
            ans = mid
            r = mid - 1
        else:
            l = mid + 1
    return ans

print("copy_time(4, 2, 3) =", copy_time(4, 2, 3))


def feed_animals(animals: list[int], food: list[int]) -> int:
    animals.sort()
    food.sort()
    count = 0
    animal_idx = 0
    for f in food:
        if animal_idx < len(animals) and f >= animals[animal_idx]:
            count += 1
            animal_idx += 1
    return count

print("feed_animals([1, 2, 3], [1, 2, 3]) =", feed_animals([1, 2, 3], [1, 2, 3]))


def find_difference_char(a: str, b: str) -> str:
    result = 0
    for char in a:
        result ^= ord(char)
    for char in b:
        result ^= ord(char)
    return chr(result)

print("find_difference_char('abcd', 'abcde') =", find_difference_char('abcd', 'abcde'))


def two_sum(data: list[int], target: int) -> list[int]:
    cache = {}
    for i, num in enumerate(data):
        diff = target - num
        if diff in cache:
            return [cache[diff], i]
        cache[num] = i
    return []

print("two_sum([2, 7, 11, 15], 9) =", two_sum([2, 7, 11, 15], 9))


from collections import defaultdict

def group_anagrams(strs: list[str]) -> list[list[str]]:
    anagram_map = defaultdict(list)
    for s in strs:
        key = tuple(sorted(s))
        anagram_map[key].append(s)
    return list(anagram_map.values())

print("group_anagrams(['eat', 'tea', 'tan', 'ate', 'nat', 'bat']) =", group_anagrams(['eat', 'tea', 'tan', 'ate', 'nat', 'bat']))
```

## Ответы на вопросы по заданиям

**1. Почему алгоритм работает за O(log n) и как определяется граница поиска?**
Алгоритм работает за O(log n), потому что на каждом шаге цикла интервал поиска делится пополам. Граница поиска определяется как `l = 0` и `r = target`, так как целая часть квадратного корня из числа `target` (при `target >= 1`) никогда не может превышать само число `target`.

**2. Как проверяется условие достаточности времени?**
Условие достаточности времени `mid` проверяется формулой `(mid // x) + (mid // y) >= n`. Она вычисляет, сколько копий сделает первый и второй ксерокс за время `mid`. Если их сумма больше или равна требуемому количеству `n`, значит, этого времени достаточно, и мы можем попробовать уменьшить правую границу поиска.

**3. Почему выбранная жадная стратегия даёт максимальное количество накормленных животных?**
Сортировка животных и порций по возрастанию гарантирует, что мы всегда пытаемся накормить самое «неприхотливое» из оставшихся животных самой маленькой подходящей порцией. Это оптимально, так как сохраняет большие порции для более «прожорливых» животных, что глобально максимизирует общее количество накормленных.

**4. Как используется XOR для нахождения лишнего элемента?**
Операция XOR (`^`) обладает свойствами: `A ^ A = 0` и `A ^ 0 = A`. Если последовательно применить XOR ко всем кодам символов (`ord`) обеих строк, то все парные символы взаимно уничтожатся (дадут 0). В результате останется только числовой код того единственного символа, который был добавлен в строку `b`. Мы преобразуем его обратно в символ через `chr()`.

**5. Как хранение уже просмотренных элементов позволяет избежать двойного цикла?**
Хранение элементов в хеш-таблице (словаре) позволяет за время O(1) проверять, встречалось ли уже нужное дополнение `target - num`. Это избавляет от необходимости запускать вложенный цикл по оставшимся элементам массива для каждого числа, снижая общую временную сложность с O(n²) до O(n).

**6. Выбранный алгоритм для группировки анаграмм и его эффективность:**
Используется хеш-таблица, где ключом является отсортированная версия строки (преобразованная в неизменяемый кортеж). Все анаграммы после сортировки дают одинаковый ключ и автоматически группируются в один список. Эффективность алгоритма составляет O(N * K log K), где N — количество строк, а K — максимальная длина строки (из-за операции сортировки каждой строки для создания ключа). Это наиболее эффективный и идиоматичный подход для данной задачи.
