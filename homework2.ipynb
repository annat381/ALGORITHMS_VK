```python
#1
def reverse_linked_list(head):
    prev = None
    current = head
    while current is not None:
        nxt = current.next      # 1) запомнить следующий
        current.next = prev     # 2) развернуть стрелку
        prev = current          # 3) сдвинуть prev
        current = nxt           # 4) сдвинуть current
    return prev                 # prev - новая голова
```

```python
#2 
def middle_node(head):
    slow = fast = head
    while fast is not None and fast.next is not None:
        slow = slow.next
        fast = fast.next.next
    return slow
```

Идея: снова slow и fast. slow идёт по 1 шагу, fast по 2. Когда fast дойдёт до конца, slow будет ровно посередине, потому что прошёл вдвое меньшее расстояние.

```python
#3
def remove_elements(head, val):
    dummy = ListNode()
    dummy.next = head
    prev = dummy
    cur = head
    while cur is not None:
        if cur.val == val:
            prev.next = cur.next   # выкидываем cur из списка
        else:
            prev = cur             # двигаем prev только если не удаляли
        cur = cur.next
    return dummy.next
```

```python
#4
def is_subsequence(a, b):
    i = j = 0
    while i < len(a) and j < len(b):
        if a[i] == b[j]:
            i += 1        # нашли очередной символ a
        j += 1            # по b двигаемся всегда
    return i == len(a)
```

i показывает, какой символ a мы сейчас ищем;
j бежит по b;
при совпадении двигаются оба, при несовпадении только j.

```python
#5
def find_pair(arr, target):
    left = 0                  # указатель в начале
    right = len(arr) - 1      # указатель в конце

    while left < right:
        current_sum = arr[left] + arr[right]

        if current_sum == target:
            return (arr[left], arr[right])   # нашли пару!
        elif current_sum < target:
            left += 1          # сумма маленькая → берём число побольше
        else:
            right -= 1         # сумма большая → берём число поменьше

    return None 

arr = [1, 3, 5, 7, 9, 11]
print(find_pair(arr, 12))   # (1, 11) или (5, 7) — зависит от порядка проверки
print(find_pair(arr, 20))   # None
```

```python
#6
def is_palindrome(s):
    stack = []
    for ch in s:
        stack.append(ch)
    for ch in s:
        if ch != stack.pop():
            return False
    return True
```

```python
#7
def remove_duplicates(arr):
    if len(arr) == 0:
        return 0

    slow = 0  # медленный указатель — позиция для записи

    for fast in range(1, len(arr)):  # быстрый бежит вперёд
        if arr[fast] != arr[slow]:   # нашли новый уникальный элемент?
            slow += 1                # двигаем медленный вперёд
            arr[slow] = arr[fast]    # перезаписываем дубликат

    return slow + 1  



arr = [1, 1, 2, 2, 2, 3, 4, 4, 5]
new_len = remove_duplicates(arr)
print(arr[:new_len])  
```

```python
#8
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val    # значение узла
        self.next = next  # ссылка на следующий узел


def make_list(arr):
    if not arr:
        return None
    head = ListNode(arr[0])
    current = head
    for val in arr[1:]:
        current.next = ListNode(val)
        current = current.next
    return head


    
def merge_two_lists(l1, l2):
    dummy = ListNode()
    tail = dummy
    while l1 is not None and l2 is not None:
        if l1.val <= l2.val:
            tail.next = l1
            l1 = l1.next
        else:
            tail.next = l2
            l2 = l2.next
        tail = tail.next
    tail.next = l1 if l1 is not None else l2   # приклеиваем остаток
    return dummy.next

    
list1 = make_list([3, 6, 8])
list2 = make_list([4, 7, 9, 11])

result = merge_two_lists(list1, list2)
print(result)  
```
